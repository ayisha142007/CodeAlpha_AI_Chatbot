```python


chatbot_responses = {
    "hello": "Hello! Welcome to CodeAlpha AI Chatbot.",
    "hi": "Hi! How can I help you?",
    "how are you": "I am fine! Thank you for asking.",
    "what is cloud computing": "Cloud computing provides computing services such as storage, servers and applications over the internet.",
    "what is python": "Python is a popular programming language used for data science, AI and software development.",
    "thank you": "You're welcome! Have a great day.",
    "bye": "Goodbye! Thank you for using the chatbot."
}

print("Predefined chatbot patterns created successfully!")
print("Number of patterns:", len(chatbot_responses))
```

```python


def chatbot(user_input):
    user_input = user_input.lower().strip()

    if user_input in chatbot_responses:
        return chatbot_responses[user_input]

    return "Sorry, I don't understand your question. Please try another query."


# Test the chatbot
print(chatbot("hello"))
print(chatbot("what is python"))
```

```python


print("====================================")
print("       CHATBOT IS READY!")
print("Type 'bye' to end the conversation.")
print("====================================")

while True:
    user_input = input("You: ")

    response = chatbot(user_input)
    print("Bot:", response)

    if user_input.lower().strip() == "bye":
        break
```

```python


test_inputs = [
    "hello",
    "hi",
    "how are you",
    "what is cloud computing",
    "what is python",
    "thank you",
    "bye",
    "unknown question"
]

print("===== CHATBOT ACCURACY TEST =====")

for user_input in test_inputs:
    response = chatbot(user_input)
    print(f"\nInput : {user_input}")
    print(f"Output: {response}")

print("\n===== TESTING COMPLETED =====")
```
